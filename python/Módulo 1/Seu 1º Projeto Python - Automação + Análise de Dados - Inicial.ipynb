```python
import pandas as pd
from unidecode import unidecode

class ProcessExcelFile:
    def __init__(self, path: str):
        self.file_path = path


    def _read_excel_file(self):
        return pd.read_excel(self.file_path)

    def _normalize_headers(self):
        db = self._read_excel_file()
        # Tudo em caixa alta, sem espaços e sem caracteres especiais
        db.columns = db.columns.map(str.upper)
        db.columns = db.columns.map(lambda x : unidecode(x.replace('.','_').replace(' ','_').strip()))
        return db

    def sum_numbers_type(self):
        db = self._normalize_headers()
        db = db.select_dtypes(include='number').sum().to_frame()
        db = db.T
        return db

class main:
    url = r"https://drive.google.com/drive/folders/14oLE59U1RqyRqlBbKpsyymW-mitvbtoh"    
    file_path = SystemNav(url=url).by_download_folder()[0]
    df = ProcessExcelFile(path= file_path).sum_numbers_type()
    display(df)
    reveneu = df['VALOR_FINAL']
    qtd_products = df['QUANTIDADE']

main()






        
```

### Vamos agora enviar um e-mail pelo gmail

```python
import smtplib
from email.mime.text import MIMEText
import locale

locale.setlocale(locale.LC_MONETARY, 'pt_BR.UTF-8')

url = r"https://drive.google.com/drive/folders/14oLE59U1RqyRqlBbKpsyymW-mitvbtoh"    
file_path = SystemNav(url=url).by_download_folder()[0]
df = ProcessExcelFile(path= file_path).sum_numbers_type()
revenue = locale.currency(df['VALOR_FINAL'].values[0])
qtd_products = df['QUANTIDADE'].values[0]

subject = "Automação Python - Relatório de Faturamento"
body = """
<html>
    <body>
    <p>Bom dia!</p>
    <br>
    Espero que encontrem-se bem.
    <br>
    <p>Segue números do relatório de vendas:</p>
    <ul>
        <li><b>Quantidade de produtos vendidos:</b> {}</li>
        <li><b>Faturamento:</b> {}</li>
    </ul>

    </body>
</html>
""".format(qtd_products,revenue)
body = MIMEText(body, 'html')
sender = "email@gmail.com"
recipients = ["email@gmail.com"]
password = ""

def send_email(subject, body, sender, recipients, password):
    msg = body
    msg['Subject'] = subject
    msg['From'] = sender
    msg['To'] = ', '.join(recipients)
    with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp_server:
       smtp_server.login(sender, password)
       smtp_server.sendmail(sender, recipients, msg.as_string())
    print("Mensagem enviada!")


send_email(subject, body, sender, recipients, password)
```
