#### 1. Tamanho de strings. Faça um programa que leia 2 strings e informe o conteúdo delas seguido do seu comprimento. Informe também se as duas strings possuem o mesmo comprimento e são iguais ou diferentes no conteúdo.
<pre>
Compara duas strings
String 1: Brasil Hexa 2006
String 2: Brasil! Hexa 2006!
Tamanho de "Brasil Hexa 2006": 16 caracteres
Tamanho de "Brasil! Hexa 2006!": 18 caracteres
As duas strings são de tamanhos diferentes.
As duas strings possuem conteúdo diferente.
</pre>

```python
class AppCompareStrings:
    def __init__(self, str_1: str, str_2: str):
        self.str_1 = str_1
        self.str_2 = str_2
        self.len_str_1 = len(str_1)
        self.len_str_2 = len(str_2)

    def _compare_content(self):
        if self.str_1 == self.str_2:
            return True
        else:
            return False

    def _str_compare_content(self):
        if self._compare_content():
            return "As duas strings possuem o mesmo conteúdo."
        else:
            return "As duas strings possuem conteúdo diferente."

    def _compare_len(self):
        if self.len_str_1 == self.len_str_2:
            return True
        else:
            return False

    def _str_compare_len(self):
        if self._compare_len():
            return "As duas strings são de mesmo tamanho."
        else:
            return "As duas strings são de tamanhos diferentes."

    def print_comparation(self):
        title_print = "Comparação das duas Strings"
        separation_bar = "="*len(title_print)
        print(f"{title_print}\n{separation_bar}\nString 1: {self.str_1}\nString 2: {self.str_2}\n{separation_bar}\nTamanho String 1: {self.len_str_1}\nTamanho String 2: {self.len_str_2}\n{separation_bar}\n{self._str_compare_len()}\n{self._str_compare_content()}")

def main():
    str_1 = "Brasil Hexa 2006"
    str_2 = "Brasil! Hexa 2006!"
    compare = AppCompareStrings(str_1=str_1,str_2=str_2)
    compare.print_comparation()
    return

main()
```

#### 2. Valida e corrige número de telefone. Faça um programa que leia um número de telefone, e corrija o número no caso deste conter somente 7 dígitos, acrescentando o '3' na frente. O usuário pode informar o número com ou sem o traço separador.
<pre>
Valida e corrige número de telefone
Telefone: 461-0133
Telefone possui 7 dígitos. Vou acrescentar o digito três na frente.
Telefone corrigido sem formatação: 34610133
Telefone corrigido com formatação: 3461-0133
</pre>

```python
import re

class PhoneValidation:
    def __init__(self, tel: str):
        self.tel = tel
        self.len_default = 8

    def _clean_input_tel(self):
        return re.sub(r'\D',"",self.tel)

    def _len_tel(self):
        return len(self._clean_input_tel())

    def _verify_len(self):
        
        if self._len_tel() == self.len_default:
            return self._clean_input_tel()
        elif self._len_tel() == self.len_default-1:
            return '3'+self._clean_input_tel()
        else:
            return False

    def _format_tel(self):
        middle = self.len_default//2
        format_tel = self._verify_len()[:middle]+'-'+self._verify_len()[middle:]
        return format_tel

    def print_result(self):
        title_print = "Valida e corrige número de telefone"
        separation_bar = "="*len(title_print)

        if self._verify_len() != False and self._len_tel() == self.len_default-1:
            print(f"""
            {title_print}
            {separation_bar}
            Telefone: {self._clean_input_tel()}
            Telefone possui {self._len_tel()} dígitos. Vou acrescentar o digito três na frente.
            Telefone corrigido sem formatação: {self._verify_len()}
            Telefone corrigido com formatação: {self._format_tel()}
            {separation_bar}
            """)

        elif self._verify_len() != False and self._len_tel() == self.len_default:
            print(f"""
            {title_print}
            {separation_bar}
            Telefone: {self._clean_input_tel()}
            Telefone possui {self._len_tel()} dígitos.
            Telefone corrigido sem formatação: {self._verify_len()}
            Telefone corrigido com formatação: {self._format_tel()}
            {separation_bar}
            """)

        else:
            print("Valor de input inválido!")

def main():
    phone = "9461-0133"
    PhoneValidation(tel=phone).print_result()
        
main()
        
```
